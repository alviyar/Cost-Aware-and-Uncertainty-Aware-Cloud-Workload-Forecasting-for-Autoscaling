```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.preprocessing import MinMaxScaler
from statsmodels.tsa.statespace.sarimax import SARIMAX
from prophet import Prophet
import xgboost as xgb
import lightgbm as lgb
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, GRU, Dense, Conv1D, MaxPooling1D, Dropout
from scipy.stats import wilcoxon

np.random.seed(42)
tf.random.set_seed(42)
```

```python
from datacentertracesdatasets import loadtraces

trace = loadtraces.get_trace(trace_name='alibaba2018', trace_type='machine_usage', stride_seconds=300)
series = trace["cpu_util_percent"].reset_index(drop=True).astype(float)
series.index = pd.date_range("2018-01-01", periods=len(series), freq="5min")
series.name = "cpu"
series.plot(figsize=(12, 3), title="Alibaba 2018 - data-center CPU %"); plt.show()
print(series.index[0], len(series))
```

```python
def make_features(s, lags=(1,2,3,6,12,288)):
    d = pd.DataFrame({"y": s})
    for l in lags:
        d[f"lag_{l}"] = s.shift(l)
    d["roll_mean_12"] = s.shift(1).rolling(12).mean()
    d["roll_std_12"] = s.shift(1).rolling(12).std()
    d["hour"] = d.index.hour
    d["dow"] = d.index.dayofweek
    return d.dropna()

data = make_features(series)
n = len(data)
train = data.iloc[:int(n*0.7)]
val   = data.iloc[int(n*0.7):int(n*0.85)]
test  = data.iloc[int(n*0.85):]

X_cols = [c for c in data.columns if c != "y"]
X_train, y_train = train[X_cols], train["y"]
X_val, y_val = val[X_cols], val["y"]
X_test, y_test = test[X_cols], test["y"]
print(len(train), len(val), len(test))
```

```python
def rmse(y, p): return np.sqrt(mean_squared_error(y, p))
def mape(y, p): return np.mean(np.abs((y - p) / np.maximum(y, 1e-6))) * 100

def provisioning_cost(actual, forecast, unit_cost=1.0, sla_penalty=5.0):
    over  = np.maximum(forecast - actual, 0)   # wasted capacity
    under = np.maximum(actual - forecast, 0)   # SLA violation
    cost = unit_cost*over.sum() + sla_penalty*under.sum()
    violation_rate = (under > 0).mean()
    return cost, violation_rate
```

```python
model = xgb.XGBRegressor(n_estimators=300, learning_rate=0.05, max_depth=5)
model.fit(X_train, y_train)
pred = model.predict(X_test)

cost, vr = provisioning_cost(y_test.values, pred)
print(f"RMSE {rmse(y_test, pred):.2f} | MAPE {mape(y_test, pred):.2f}% | cost {cost:.0f} | SLA violations {vr:.1%}")
```

```python
# Upper-bound forecast: 90th percentile with LightGBM
q = 0.90
qmodel = lgb.LGBMRegressor(objective="quantile", alpha=q, n_estimators=300, learning_rate=0.05)
qmodel.fit(X_train, y_train)
pred_q = qmodel.predict(X_test)

cost_q, vr_q = provisioning_cost(y_test.values, pred_q)
print(f"Point : cost {cost:.0f} | violations {vr:.1%}")
print(f"Q{int(q*100)}   : cost {cost_q:.0f} | violations {vr_q:.1%}")
```

```python
qs = [0.5, 0.7, 0.8, 0.9, 0.95, 0.99]
rows = []
for q in qs:
    m = lgb.LGBMRegressor(objective="quantile", alpha=q, n_estimators=300, learning_rate=0.05, verbose=-1)
    m.fit(X_train, y_train)
    p = m.predict(X_test)
    c, v = provisioning_cost(y_test.values, p)
    rows.append((q, c, v))
qdf = pd.DataFrame(rows, columns=["quantile", "cost", "violation_rate"])
print(qdf)

fig, ax = plt.subplots(1, 2, figsize=(10, 3))
ax[0].plot(qdf.quantile if False else qdf["quantile"], qdf["cost"], marker="o"); ax[0].set_title("Total cost vs quantile")
ax[1].plot(qdf["quantile"], qdf["violation_rate"], marker="o"); ax[1].set_title("SLA violation rate vs quantile")
plt.show()
```

```python
W = 24  # look-back window (24 x 5 min = 2 hours)
pos = lambda idx: series.index.get_loc(idx)
tr_s, va_s, te_s = pos(train.index[0]), pos(val.index[0]), pos(test.index[0])
te_e = pos(test.index[-1]) + 1

scaler = MinMaxScaler().fit(series.iloc[:va_s].values.reshape(-1, 1))  # fit on train only
arr = scaler.transform(series.values.reshape(-1, 1)).ravel()

def windows(start, end):
    X = np.array([arr[i-W:i] for i in range(start, end)])[..., None]
    y = np.array([arr[i] for i in range(start, end)])
    return X, y

Xtr, ytr = windows(tr_s, va_s)
Xva, yva = windows(va_s, te_s)
Xte, yte = windows(te_s, te_e)

def build(kind):
    m = Sequential()
    if kind == "LSTM":
        m.add(LSTM(64, input_shape=(W, 1)))
    elif kind == "GRU":
        m.add(GRU(64, input_shape=(W, 1)))
    else:  # CNN-LSTM
        m.add(Conv1D(32, 3, activation="relu", input_shape=(W, 1)))
        m.add(MaxPooling1D(2))
        m.add(LSTM(64))
    m.add(Dropout(0.1)); m.add(Dense(1))
    m.compile(optimizer="adam", loss="mse")
    return m

deep_preds = {}
es = tf.keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True)
for kind in ["LSTM", "GRU", "CNN-LSTM"]:
    m = build(kind)
    m.fit(Xtr, ytr, validation_data=(Xva, yva), epochs=25, batch_size=64, callbacks=[es], verbose=0)
    p = m.predict(Xte, verbose=0).ravel()
    deep_preds[kind] = scaler.inverse_transform(p.reshape(-1, 1)).ravel()
    print(kind, "done")
```

```python
y_true = y_test.values
all_preds = {"XGBoost": model.predict(X_test), **deep_preds}

res = []
for name, p in all_preds.items():
    c, v = provisioning_cost(y_true, p)
    res.append([name, rmse(y_true, p), mape(y_true, p), c, v])

# add quantile-based provisioning (point forecast vs safety margin)
m90 = lgb.LGBMRegressor(objective="quantile", alpha=0.9, n_estimators=300, learning_rate=0.05, verbose=-1).fit(X_train, y_train)
c, v = provisioning_cost(y_true, m90.predict(X_test))
res.append(["LightGBM Q90", rmse(y_true, m90.predict(X_test)), mape(y_true, m90.predict(X_test)), c, v])

results = pd.DataFrame(res, columns=["Model", "RMSE", "MAPE %", "Total cost", "SLA violation rate"]).sort_values("Total cost")
print(results.round(2).to_string(index=False))
```

```python
# compare absolute errors of two models on the same test points
e1 = np.abs(y_true - all_preds["XGBoost"])
e2 = np.abs(y_true - all_preds["LSTM"])
print(wilcoxon(e1, e2))
```

```python
# store point predictions and quantile predictions
q_models = {}
for q in [0.5, 0.7, 0.8, 0.9, 0.95, 0.99]:
    q_models[q] = lgb.LGBMRegressor(objective="quantile", alpha=q, n_estimators=300,
                                     learning_rate=0.05, verbose=-1).fit(X_train, y_train)

q_preds = {f"LGBM Q{int(q*100)}": m.predict(X_test) for q, m in q_models.items()}
candidates = {**all_preds, **q_preds}

penalties = [1, 2, 5, 10, 20]
sens = pd.DataFrame(index=candidates.keys(), columns=penalties, dtype=float)
for name, p in candidates.items():
    for pen in penalties:
        sens.loc[name, pen] = provisioning_cost(y_true, p, sla_penalty=pen)[0]

print(sens.round(0))
print("\nCheapest model per penalty ratio:")
print(sens.idxmin())
```

```python
def conformal_upper(q, X_cal, y_cal, X_new, base_model):
    """Shift the quantile forecast so coverage on new data is about q."""
    scores = y_cal.values - base_model.predict(X_cal)          # positive = under-predicted
    n_cal = len(scores)
    level = min(np.ceil((n_cal + 1) * q) / n_cal, 1.0)
    shift = np.quantile(scores, level)
    return base_model.predict(X_new) + shift

q = 0.9
p_conf = conformal_upper(q, X_val, y_val, X_test, q_models[0.9])
p_raw = q_models[0.9].predict(X_test)

for label, p in [("Raw Q90", p_raw), ("Conformal Q90", p_conf)]:
    c, v = provisioning_cost(y_true, p)
    print(f"{label:14s} | cost {c:.0f} | violations {v:.1%} | coverage {(1-v):.1%}")
```

```python
train_part = series.iloc[max(0, va_s - 3000):va_s]          # last 3000 training points keeps it fast
arima = SARIMAX(train_part, order=(2, 0, 1), trend="c").fit(disp=False)

# apply the fitted model to the full series, then read one-step-ahead predictions for the test span
full = arima.apply(series.iloc[:te_e], refit=False)
pred_arima = full.get_prediction(start=te_s, end=te_e - 1).predicted_mean.values

c, v = provisioning_cost(y_true, pred_arima)
print(f"ARIMA | RMSE {rmse(y_true, pred_arima):.2f} | MAPE {mape(y_true, pred_arima):.2f}% | cost {c:.0f} | violations {v:.1%}")
all_preds["ARIMA"] = pred_arima
```

```python
# Naive baseline: predict that the next value equals the previous one (lag_1)
pred_naive = X_test["lag_1"].values
c, v = provisioning_cost(y_true, pred_naive)
print(f"Naive | RMSE {rmse(y_true, pred_naive):.2f} | MAPE {mape(y_true, pred_naive):.2f}% | cost {c:.0f} | violations {v:.1%}")
all_preds["Naive"] = pred_naive
```

```python
# ---- Rebuild all predictions (Cell 9 overwrote all_preds, so ARIMA/Naive were dropped) ----
y_true = y_test.values
all_preds = {
    "Naive": X_test["lag_1"].values,
    "ARIMA": pred_arima,
    "XGBoost": model.predict(X_test),
    **deep_preds,
}

# Quantile models (point forecast = Q50, safety margins = higher quantiles)
q_models = {}
for q in [0.5, 0.7, 0.8, 0.9, 0.95, 0.99]:
    q_models[q] = lgb.LGBMRegressor(objective="quantile", alpha=q, n_estimators=300,
                                     learning_rate=0.05, verbose=-1).fit(X_train, y_train)
q_preds = {f"LGBM Q{int(q*100)}": m.predict(X_test) for q, m in q_models.items()}
candidates = {**all_preds, **q_preds}

# ---- Table 1: accuracy and cost at penalty 5 ----
rows = []
for name, p in candidates.items():
    c, v = provisioning_cost(y_true, p, sla_penalty=5)
    rows.append([name, rmse(y_true, p), mape(y_true, p), c, v])
table1 = pd.DataFrame(rows, columns=["Model", "RMSE", "MAPE %", "Total cost", "SLA violation rate"])
print("=== Accuracy vs cost (penalty = 5x) ===")
print(table1.sort_values("Total cost").round(2).to_string(index=False))

# ---- Table 2: penalty sensitivity ----
penalties = [1, 2, 5, 10, 20]
sens = pd.DataFrame(index=candidates.keys(), columns=penalties, dtype=float)
for name, p in candidates.items():
    for pen in penalties:
        sens.loc[name, pen] = provisioning_cost(y_true, p, sla_penalty=pen)[0]
print("\n=== Total cost by SLA penalty ratio ===")
print(sens.round(0))
print("\nCheapest model per penalty:")
print(sens.idxmin())

# ---- Table 3: raw vs conformal Q90 ----
p_raw = q_models[0.9].predict(X_test)
p_conf = conformal_upper(0.9, X_val, y_val, X_test, q_models[0.9])
print("\n=== Q90: raw vs conformal ===")
for label, p in [("Raw Q90", p_raw), ("Conformal Q90", p_conf)]:
    c, v = provisioning_cost(y_true, p)
    print(f"{label:14s} | cost {c:.0f} | violations {v:.1%} | coverage {1-v:.1%}")
```

```python
from datacentertracesdatasets import loadtraces
from scipy.stats import wilcoxon

# ---------------- settings ----------------
TRACES    = ["alibaba2018", "azure_v2", "google2019"]
N_WINDOWS = 3            # test windows per trace (each = 1 day = 288 points)
TEST_LEN  = 288
SEEDS     = [0, 1]       # seeds for deep models
DEEP      = False        # first run False; then True for the full run
PENALTIES = [1, 2, 5, 10, 20]
QUANTILES = [0.5, 0.7, 0.8, 0.9, 0.95, 0.99]
W = 24

def load_series(name):
    tr = loadtraces.get_trace(trace_name=name, trace_type='machine_usage', stride_seconds=300)
    if isinstance(tr, pd.DataFrame):
        col = next((c for c in tr.columns if "cpu" in str(c).lower()), tr.columns[0])
        s = tr[col].astype(float).reset_index(drop=True)
        print(f"  {name}: using column '{col}' of {list(tr.columns)}")
    else:
        arr = np.asarray(tr)
        if arr.ndim == 3: arr = arr.reshape(-1, arr.shape[-1])
        s = pd.Series(arr[:, 0].astype(float))
        print(f"  {name}: ndarray {arr.shape}, using column 0 (CHECK this is CPU)")
    s.index = pd.date_range("2018-01-01", periods=len(s), freq="5min")
    return s.interpolate().ffill().bfill()

def build_deep(kind):
    m = Sequential()
    if kind == "LSTM":   m.add(LSTM(64, input_shape=(W, 1)))
    elif kind == "GRU":  m.add(GRU(64, input_shape=(W, 1)))
    else:
        m.add(Conv1D(32, 3, activation="relu", input_shape=(W, 1))); m.add(MaxPooling1D(2)); m.add(LSTM(64))
    m.add(Dropout(0.1)); m.add(Dense(1)); m.compile(optimizer="adam", loss="mse")
    return m

def deep_predict(kind, s, tr_pos, va_pos, te_pos, seed):
    tf.keras.utils.set_random_seed(seed)
    sc = MinMaxScaler().fit(s.values[tr_pos].reshape(-1, 1))
    arr = sc.transform(s.values.reshape(-1, 1)).ravel()
    mk = lambda pos: (np.array([arr[i-W:i] for i in pos])[..., None], arr[pos])
    Xa, ya = mk(tr_pos); Xb, yb = mk(va_pos); Xc, _ = mk(te_pos)
    m = build_deep(kind)
    m.fit(Xa, ya, validation_data=(Xb, yb), epochs=25, batch_size=64, verbose=0,
          callbacks=[tf.keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True)])
    p = m.predict(Xc, verbose=0).ravel()
    return sc.inverse_transform(p.reshape(-1, 1)).ravel()

def metrics_row(name, k, model, y, p):
    row = dict(trace=name, window=k, model=model, rmse=rmse(y, p))
    for pen in PENALTIES:
        row[f"cost_{pen}"] = provisioning_cost(y, p, sla_penalty=pen)[0]
    row["viol_5"] = provisioning_cost(y, p, sla_penalty=5)[1]
    return row

# ---------------- rolling-origin backtest ----------------
records = []
for name in TRACES:
    try:
        s = load_series(name)
    except Exception as e:
        print("  skipping", name, "->", e); continue
    data = make_features(s)
    pos_all = s.index.get_indexer(data.index)
    n = len(data)
    print(f"{name}: {len(s)} points, {n} usable rows")
    cols = [c for c in data.columns if c != "y"]

    for k in range(N_WINDOWS):
        a = n - (N_WINDOWS - k) * TEST_LEN
        if a < 800:
            print("  window", k, "skipped (series too short)"); continue
        trv, te = data.iloc[:a], data.iloc[a:a + TEST_LEN]
        cut = int(len(trv) * 0.85)
        tr, va = trv.iloc[:cut], trv.iloc[cut:]
        y = te["y"].values
        tr_pos, va_pos, te_pos = pos_all[:cut], pos_all[cut:a], pos_all[a:a + TEST_LEN]

        recs = [metrics_row(name, k, "Naive", y, te["lag_1"].values)]

        # ARIMA (one-step-ahead, fitted on the training part only)
        try:
            ar = SARIMAX(s.iloc[max(0, tr_pos[-1] - 2000):tr_pos[-1] + 1], order=(2, 0, 1), trend="c").fit(disp=False)
            full = ar.apply(s.iloc[:te_pos[-1] + 1], refit=False)
            p_ar = full.get_prediction(start=te_pos[0], end=te_pos[-1]).predicted_mean.values
            recs.append(metrics_row(name, k, "ARIMA", y, p_ar))
        except Exception as e:
            print("  ARIMA failed:", e)

        # XGBoost
        xg = xgb.XGBRegressor(n_estimators=300, learning_rate=0.05, max_depth=5).fit(tr[cols], tr["y"])
        recs.append(metrics_row(name, k, "XGBoost", y, xg.predict(te[cols])))

        # LightGBM quantiles + conformal Q90
        qm = {}
        for q in QUANTILES:
            qm[q] = lgb.LGBMRegressor(objective="quantile", alpha=q, n_estimators=300,
                                      learning_rate=0.05, verbose=-1).fit(tr[cols], tr["y"])
            recs.append(metrics_row(name, k, f"LGBM Q{int(q*100)}", y, qm[q].predict(te[cols])))
        p_conf = conformal_upper(0.9, va[cols], va["y"], te[cols], qm[0.9])
        recs.append(metrics_row(name, k, "LGBM Q90 conformal", y, p_conf))

        # Deep models (metrics averaged over seeds)
        if DEEP:
            for kind in ["LSTM", "GRU", "CNN-LSTM"]:
                per_seed = pd.DataFrame([metrics_row(name, k, kind, y, deep_predict(kind, s, tr_pos, va_pos, te_pos, sd))
                                         for sd in SEEDS])
                row = per_seed.iloc[0].copy()
                num = per_seed.select_dtypes("number").columns
                row[num] = per_seed[num].mean()
                recs.append(row.to_dict())

        records += recs
        print(f"  window {k} done")

res = pd.DataFrame(records)
res.to_csv("backtest_results.csv", index=False)

# ---------------- relative to Naive, then average ----------------
base = res[res.model == "Naive"].set_index(["trace", "window"])
for pen in PENALTIES:
    res[f"rel_{pen}"] = res.apply(lambda r: r[f"cost_{pen}"] / base.loc[(r.trace, r.window), f"cost_{pen}"], axis=1)
res["rel_rmse"] = res.apply(lambda r: r.rmse / base.loc[(r.trace, r.window), "rmse"], axis=1)

agg = {"rel_rmse": "mean", "viol_5": "mean", **{f"rel_{p}": "mean" for p in PENALTIES}}
summary = res.groupby("model").agg(agg).sort_values("rel_5")
print("\n=== Mean over all traces and windows (cost relative to Naive; below 1.00 = cheaper than Naive) ===")
print(summary.round(2))

print("\nCheapest model per penalty (mean relative cost):")
print(summary[[f"rel_{p}" for p in PENALTIES]].idxmin())

# ---------------- paired tests across (trace, window) pairs ----------------
piv = res.pivot_table(index=["trace", "window"], columns="model", values="rel_5")
print(f"\nPaired windows available: {len(piv)}")
for a_, b_ in [("XGBoost", "LGBM Q90"), ("LGBM Q90", "LGBM Q90 conformal")]:
    if a_ in piv and b_ in piv:
        d = (piv[a_] - piv[b_]).dropna()
        print(f"{a_} vs {b_} (penalty 5): mean diff {d.mean():+.3f}, wins for {b_}: {(d > 0).sum()}/{len(d)}, "
              f"Wilcoxon p = {wilcoxon(piv[a_].loc[d.index], piv[b_].loc[d.index]).pvalue:.4f}")
```

```python
import os, zipfile
from scipy.stats import wilcoxon

# ---------- 1) Clean LaTeX main table (best value per column in bold) ----------
cols = {"rel_rmse": "Rel. RMSE", "rel_1": "1x", "rel_2": "2x", "rel_5": "5x", "rel_10": "10x", "rel_20": "20x"}
t = summary[list(cols)].copy()
t["viol"] = summary["viol_5"] * 100
t = t.sort_values("rel_5")
best = {c: t[c].min() for c in cols}

def fmt(c, v):
    s = f"{v:.2f}"
    return f"\\textbf{{{s}}}" if abs(v - best[c]) < 1e-9 else s

lines = []
for m, r in t.iterrows():
    cells = [m] + [fmt(c, r[c]) for c in cols] + [f"{r['viol']:.1f}"]
    lines.append(" & ".join(cells) + r" \\")
latex = (
    "\\begin{table*}[t]\n\\centering\n"
    "\\caption{Mean cost relative to the Naive baseline over 3 traces and 9 test windows "
    "(values below 1.00 are cheaper than Naive; best per column in bold). "
    "Violation rate is for one-step forecasts and is independent of the penalty.}\n"
    "\\label{tab:main}\n\\begin{tabular}{lcccccc|c}\n\\toprule\n"
    " & Rel. RMSE & \\multicolumn{5}{c|}{Relative cost at penalty ratio} & SLA viol. (\\%) \\\\\n"
    " & & 1x & 2x & 5x & 10x & 20x & \\\\\n\\midrule\n"
    + "\n".join(lines) + "\n\\bottomrule\n\\end{tabular}\n\\end{table*}\n"
)
print(latex)

# ---------- 2) Paired tests across the (trace, window) pairs ----------
def paired(pen, a_, b_):
    piv = res.pivot_table(index=["trace", "window"], columns="model", values=f"rel_{pen}")
    if a_ not in piv or b_ not in piv: return
    d = (piv[a_] - piv[b_]).dropna()
    try:    p = wilcoxon(piv[a_].loc[d.index], piv[b_].loc[d.index]).pvalue
    except: p = float("nan")
    print(f"penalty {pen:>2}x | {a_:>10} vs {b_:<18} | mean diff {d.mean():+.3f} | "
          f"{b_} cheaper in {(d > 0).sum()}/{len(d)} | p = {p:.4f}")

print("\n=== Paired tests ===")
for pen in [5, 10]:
    for a_, b_ in [("XGBoost", "LGBM Q90"), ("CNN-LSTM", "LGBM Q90"), ("GRU", "LGBM Q90"),
                   ("LSTM", "LGBM Q90"), ("XGBoost", "CNN-LSTM"), ("LGBM Q90", "LGBM Q90 conformal")]:
        paired(pen, a_, b_)

# ---------- 3) Per-trace table (penalty 5x) ----------
per_trace = res.groupby(["trace", "model"])["rel_5"].mean().unstack(0).round(2)
per_trace = per_trace.loc[summary.index]
print("\n=== Relative cost at 5x, per trace ===")
print(per_trace)

# ---------- 4) Save and download everything ----------
with open("table_main.tex", "w") as f: f.write(latex)
summary.round(3).to_csv("table_summary.csv")
per_trace.to_csv("table_per_trace.csv")
res.to_csv("backtest_results.csv", index=False)

outs = ["table_main.tex", "table_summary.csv", "table_per_trace.csv", "backtest_results.csv"] + \
       [f for f in os.listdir(".") if f.startswith("fig") and f.endswith(".png")]
with zipfile.ZipFile("paper_outputs.zip", "w") as z:
    for f in outs: z.write(f)
print("\nSaved:", outs)

try:
    from google.colab import files
    files.download("paper_outputs.zip")
except Exception as e:
    print("Download manually from the Colab file panel:", e)
```
